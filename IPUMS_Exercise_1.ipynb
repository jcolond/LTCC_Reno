```r
library(ipumsr)
library(dplyr)
```

**PART 1**

```r
ddi <- read_ipums_ddi ("usa_00071.xml")
data <- read_ipums_micro(ddi)
```

```r
data$FARM
```

```r
data %>% group_by(YEAR, FARM = haven::as_factor(FARM, level = "both")) %>% summarize(n = sum(PERWT)) %>% mutate (pct = n/sum(n))
```

```r
data %>% filter(PERNUM == 1 & YEAR == 1940) %>% group_by(FARM = haven::as_factor(FARM)) %>% summarize (n = sum(PERNUM)) %>% mutate (pct = n/sum(n))
```

```r
data %>% filter(PERNUM == 1 & YEAR == 1940) %>% group_by(FARM = haven::as_factor(FARM)) %>% summarize (n = sum(HHWT)) %>% mutate (pct = n / sum(n))
```

**PART 2**

```r
ddi <- read_ipums_ddi ("usa_00072.xml")
data <- read_ipums_micro(ddi)
```

```r
data %>% filter(YEAR == 1980) %>% group_by(VETSTAT = haven::as_factor(VETSTAT), EMPSTAT = haven::as_factor(EMPSTAT)) %>% summarize(n = sum(PERWT)) %>% mutate(pct = n/sum(n))
```

```r
data %>% filter(YEAR == 2000) %>% group_by(VETSTAT = haven::as_factor(VETSTAT), EMPSTAT = haven::as_factor(EMPSTAT)) %>% summarize(n = sum(PERWT)) %>% mutate(pct = n/sum(n))
```

```r
data %>% filter(YEAR == 2000 & EMPSTAT != 3) %>% group_by(VETSTAT = haven::as_factor(VETSTAT), EMPSTAT = haven::as_factor(EMPSTAT)) %>% summarize(n = sum(PERWT)) %>% mutate (pct= n/sum(n))
```

**PART 3**

```r
ddi <- read_ipums_ddi ("usa_00073.xml")
data <- read_ipums_micro(ddi)
```

```r
data <- data %>% mutate(CAR = lbl_relabel(CARPOOL, lbl(2, "Any form of carpooling") ~ .val %in% c(2, 3, 4, 5)))
```

```r
data %>% filter(YEAR == 1980 & METRO %in% c(1, 3, 5)) %>% group_by(METRO = haven::as_factor(METRO), CAR = haven::as_factor(CAR)) %>% summarize(n = sum(PERWT)) %>% mutate(pct = n/sum(n))
```

```r
data %>% filter(YEAR == 2010 & METRO %in% c(1, 3, 5)) %>% group_by(METRO = haven::as_factor(METRO), CAR = haven::as_factor(CAR)) %>% summarize(n = sum(PERWT)) %>% mutate(pct = n/sum(n))
```
